# Data Visualizations

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Top 10 Run Scorers
top_runs = df.groupby("Player_name")["Total_runs"].sum().sort_values(ascending=False).head(10)

# Bar Plot
top_runs.plot(kind="bar", color="#27DDF5")

plt.title("Top 10 Run Scorers")
plt.xlabel("Player Name")
plt.ylabel("Total Runs")
plt.xticks(rotation=20)
plt.show()
```

```python
# Top 10 Wicket Takers:
top_wickets = df.groupby("Player_name")["Total_wickets_taken"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
plt.barh(top_wickets.index, top_wickets.values)

plt.title("Top 10 Wicket Takers")
plt.xlabel("Total Wickets")
plt.ylabel("Player")
plt.gca().invert_yaxis()
plt.show()
```

```python
#Batting Average vs Strike Rate:
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x="Average",
    y="Strike_rate",
    size="Total_runs",
    alpha=0.6,
    color="purple"
)

plt.title("Batting Average vs Strike Rate")
plt.xlabel("Batting Average")
plt.ylabel("Strike Rate")
plt.show()
```

```python
#Runs Distribution by Team:
plt.figure(figsize=(12,6))

sns.boxplot(
    data=df,
    x="Team",
    y="Total_runs",
    color="blue"
)

plt.title("Runs Distribution by Team")
plt.xlabel("Team")
plt.ylabel("Total Runs")
plt.xticks(rotation=45)
plt.show()
```

```python
#India vs Australia - Total Runs:
teams = df[df["Team"].isin(["India", "Australia"])]

team_runs = teams.groupby("Team")["Total_runs"].sum()

plt.figure(figsize=(6,6))

plt.pie(
    team_runs,
    labels=team_runs.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=["purple", "green"]
)

plt.title("India vs Australia - Total Runs")
plt.show()
```

```python
#Top Players Run Comparison:
top_players = df.groupby("Player_name")["Total_runs"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))

plt.plot(
    top_players.index,
    top_players.values,
    marker="*"
)

plt.title("Top Players Run Comparison")
plt.xlabel("Player")
plt.ylabel("Total Runs")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
#Runs vs Wickets:

plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Total_runs",
    y="Total_wickets_taken"
)

plt.title("Runs vs Wickets")
plt.xlabel("Total Runs")
plt.ylabel("Total Wickets")
plt.show()
```

```python
#Number of Players by Team:

plt.figure(figsize=(10,6))
sns.countplot(
    data=df,
    x="Team"
)

plt.title("Number of Players by Team")
plt.xlabel("Team")
plt.ylabel("Number of Players")
plt.xticks(rotation=45)

plt.show()
```

```python
#All-Rounder Performance:

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df,
    x="Total_runs",
    y="Total_wickets_taken",
    alpha=0.6
)

plt.title("All-Rounder Performance")
plt.xlabel("Total Runs")
plt.ylabel("Total Wickets")

plt.tight_layout()
plt.show()
```

```python
#Cricket Player Performance Correlation:
columns = [
    "Total_runs",
    "Strike_rate",
    "Total_wickets_taken",
    "Total_runs_conceded",
    "Total_overs_bowled",
    "Total_matches_played",
    "Matches_won",
    "Matches_lost",
    "Average"
]

plt.figure(figsize=(10,7))

sns.heatmap(
    df[columns].corr(),
    annot=True,
    fmt=".2f"
)

plt.title("Cricket Player Performance Correlation")
plt.show()
```
